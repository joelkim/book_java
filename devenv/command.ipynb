# 자바 명령어

- JDK는 `jar`, `javadoc`, `jshell`, `jdeps` 등 다양한 개발 도구를 제공하지만, **소스의 컴파일과 실행**에 사용하는 핵심 명령어는 다음 2개임
  - `javac`: 소스 파일(`.java`)을 바이트코드(`.class`)로 **컴파일**
  - `java` : 컴파일된 바이트코드(또는 단일 소스 파일)를 JVM에서 **실행**

## javac 컴파일 명령

- 인수로 컴파일할 소스 파일(`.java`)의 경로를 받음
- 컴파일에 성공하면 같은 이름의 `.class` 파일(바이트코드)이 생성됨
- 출력 위치를 지정하지 않으면 소스 파일과 같은 위치에 `.class`가 생성됨

```bash
# 기본 컴파일 → Hello.class 생성
javac Hello.java

# 여러 파일 동시 컴파일
javac Hello.java World.java

# -d 옵션으로 출력 디렉터리(클래스 루트) 지정
javac -d out Hello.java
```

> 참고: 패키지가 선언된 소스를 `-d`로 컴파일하면 패키지 경로에 맞춰 하위 디렉터리가 자동 생성됨
> (예: `package com.example;` → `out/com/example/Hello.class`)

## java 실행 명령

- 인수로 실행할 **클래스 이름**을 받음 (파일 이름이나 `.class` 확장자가 아님)
- 지정한 클래스에는 진입점인 `public static void main(String[] args)` 메서드가 있어야 함
- 실행할 클래스가 패키지에 포함되는 경우, 패키지 경로를 포함한 **완전 정규화된 클래스 이름(FQCN, Fully Qualified Class Name)** 을 주어야 함

```bash
# 패키지가 없는 경우
java Hello

# 패키지에 포함된 경우 → 점(.)으로 구분한 전체 이름 사용
java com.example.Hello
```

- **Java 11 이상**에서는 `javac` 컴파일 없이 단일 소스 파일을 직접 실행할 수 있음 (Single-File Source-Code Program)

```bash
# 내부적으로 메모리에서 컴파일 후 바로 실행 (.class 파일 생성 안 함)
java Hello.java
```

## 클래스패스와 `-cp`

- **클래스패스(classpath)** 란 자바가 클래스(`.class`)나 패키지를 찾기 시작하는 **루트 경로**를 의미함
- `-cp`, `-classpath`, `--class-path` 는 모두 같은 옵션임
- 지정하지 않으면 **현재 디렉터리(`.`)** 가 기본 클래스패스가 됨
- 단, `-cp`를 **명시하면 현재 디렉터리는 자동 포함되지 않음**. 현재 디렉터리도 함께 쓰려면 `-cp out;.`(Windows) / `-cp out:.`(Unix)처럼 직접 추가해야 함

용도
- `javac` : 컴파일할 소스가 **다른 패키지의 클래스를 참조**하는 경우, 그 클래스의 **클래스패스 루트**를 `-cp`로 지정
- `java` : 실행할 클래스가 속한 패키지의 **클래스패스 루트**를 `-cp`로 지정

```bash
# out 디렉터리를 루트로 삼아 com.example.Hello 실행
java -cp out com.example.Hello

# 여러 경로 지정 (구분자: Windows는 ;  ·  Linux/macOS는 :)
java -cp out;lib\util.jar com.example.Hello   # Windows
java -cp out:lib/util.jar com.example.Hello   # Linux/macOS
```

> 핵심: `-cp`에는 패키지 경로(`com/example`)가 아니라, 그 패키지가 **시작되는 상위 디렉터리(루트)** 를 지정해야 함
> 예) `out/com/example/Hello.class` 를 실행할 때 → 클래스 이름은 `com.example.Hello`, 클래스패스는 `out`
>
> jar 파일 자체도 클래스패스 항목으로 지정할 수 있음

## 전체 흐름 예시

`com.example.Hello` 클래스를 컴파일하고 실행하는 과정

```bash
# 디렉터리 구조
#   src/com/example/Hello.java   (package com.example;)

# 1) 컴파일: 소스를 out 디렉터리에 바이트코드로 생성
javac -d out src/com/example/Hello.java
#   → out/com/example/Hello.class 생성

# 2) 실행: out을 클래스패스 루트로, FQCN으로 클래스 지정
java -cp out com.example.Hello
```

| 명령 | 인수로 주는 것 | 예시 |
|------|---------------|------|
| `javac` | 소스 **파일 경로** | `src/com/example/Hello.java` |
| `java`  | **클래스 이름**(FQCN) | `com.example.Hello` |
| `-cp`   | 패키지가 시작되는 **루트 디렉터리** | `out` |
